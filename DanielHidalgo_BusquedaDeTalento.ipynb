# 3. Consideraciones éticas y anonimización de los CVs sintéticos

La actividad requiere generar 20 CVs sintéticos con inteligencia artificial para construir la base de conocimiento del sistema. Sin embargo, estos CVs no deben incluir datos personales sensibles ni información que permita identificar directa o indirectamente a una persona real.

Por esa razón, en esta propuesta se decidió utilizar identificadores anónimos para cada candidato, por ejemplo: `CAND-001`, `CAND-002`, `CAND-003`, etc. De esta manera, el sistema puede analizar formación, experiencia, habilidades técnicas, habilidades blandas, certificaciones, idiomas y proyectos relevantes sin depender de información personal.

## 3.1 Datos excluidos deliberadamente

En los CVs sintéticos no se incluirán los siguientes elementos:

| Dato excluido | Razón ética |
|---|---|
| Nombre de la persona | Puede generar sesgos asociados a origen social, género, cultura o familiaridad. |
| Fotografía | Puede inducir sesgos por apariencia, edad percibida, género, etnicidad o discapacidad visible. |
| Género | No es necesario para evaluar competencias profesionales y puede provocar discriminación. |
| Edad | Puede inducir sesgos por juventud o seniority percibido, independientemente de la experiencia real. |
| Nacionalidad | Puede generar sesgos culturales o migratorios no relacionados con el desempeño profesional. |
| Estado civil | No tiene relevancia técnica para el puesto y puede generar discriminación. |
| Dirección personal | No aporta valor al análisis semántico y compromete privacidad. |
| Teléfono o correo real | No es necesario para la actividad y podría confundirse con información de contacto real. |

## 3.2 Importancia de la anonimización

La anonimización permite que el sistema se concentre en los elementos realmente relevantes para la selección de talento: experiencia profesional, competencias técnicas, nivel de dominio, certificaciones, proyectos, herramientas utilizadas y habilidades de colaboración. Esto reduce el riesgo de que el modelo o el evaluador humano tomen decisiones basadas en atributos personales que no deberían influir en la recomendación.

Además, al usar CVs sintéticos y anónimos, se protege la privacidad de personas reales y se evita exponer datos personales en una actividad académica. Este enfoque es coherente con el propósito del sistema: apoyar el análisis inicial de candidatos sin reemplazar al reclutador ni eliminar la supervisión humana.

## 3.3 Criterio adoptado para esta actividad

Para esta actividad, cada CV será identificado únicamente mediante un código anónimo. La información incluida se limitará a:

- Resumen profesional.
- Formación académica.
- Experiencia laboral.
- Habilidades técnicas.
- Habilidades blandas.
- Certificaciones.
- Idiomas.
- Proyectos relevantes.

Con esto se busca construir una base de conocimiento útil para la búsqueda semántica, pero sin introducir datos personales innecesarios ni variables que puedan favorecer decisiones sesgadas.

# 5. Tipos de perfiles generados

A partir del diseño de la base de CVs sintéticos, se generaron perfiles variados relacionados en distinto grado con la vacante de Lead Machine Learning Engineer. Aunque algunos candidatos tienen denominaciones más específicas, como Lead Machine Learning Engineer, Data Architect o Software Architect, para el análisis general se agruparon dentro de seis categorías principales.

Esta agrupación permite mantener una diversidad controlada y evita que la base de conocimiento quede demasiado dispersa. El objetivo es contar con perfiles altamente alineados con la vacante, perfiles relacionados y perfiles parcialmente relacionados, para evaluar posteriormente si el sistema de búsqueda semántica logra jerarquizar adecuadamente a los mejores candidatos.

```python
# Agrupación de perfiles específicos en las seis categorías principales
df_diseno_cvs["categoria_perfil"] = df_diseno_cvs["perfil"].replace({
    "Lead Machine Learning Engineer": "Machine Learning Engineer",
    "Data Architect": "Data Engineer",
    "Software Architect": "Software Engineer"
})


# Se construye una tabla resumen para contar cuántos candidatos hay en cada categoría principal de perfil
df_tipos_perfil = (
    df_diseno_cvs
    .groupby("categoria_perfil")
    .agg(
        cantidad=("id", "count"),
        candidatos=("id", lambda x: ", ".join(x)),
        niveles=("nivel", lambda x: ", ".join(sorted(x.unique()))),
        formatos=("formato", lambda x: ", ".join(sorted(x.unique())))
    )
    .reset_index()
    .sort_values("cantidad", ascending=False)
)

df_tipos_perfil
```

## 6.1 Exportación de los CVs sintéticos

A partir del DataFrame `df_cvs`, se exportaron los 20 CVs sintéticos como documentos individuales. Para respetar la distribución definida en la actividad, se utilizaron 16 archivos en formato PDF y 4 archivos en formato PNG.

Estos documentos integrarán la base de conocimiento que será utilizada posteriormente para la extracción de texto, construcción de embeddings, búsqueda semántica y ranking de candidatos.


```python
import os
import textwrap
from reportlab.lib.pagesizes import letter
from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer
from reportlab.lib.styles import getSampleStyleSheet
from PIL import Image, ImageDraw, ImageFont

# Crear carpeta de salida
carpeta_cvs = "base_conocimiento_cvs"
os.makedirs(carpeta_cvs, exist_ok=True)

# Unir el contenido completo de los CVs con el diseño de formatos
df_cvs_export = df_cvs.merge(
    df_diseno_cvs[["id", "formato", "alineacion_esperada", "experiencia_anios"]],
    on="id",
    how="left"
)

# Esta función construye el texto completo de cada CV.
# Recibe una fila del DataFrame y organiza la información en formato de CV.
def construir_texto_cv(fila):
    texto = f"""
CV SINTÉTICO - {fila['id']}

Nivel: {fila['nivel']}
Perfil: {fila['perfil']}
Años de experiencia: {fila['experiencia_anios']}
Alineación esperada con la vacante: {fila['alineacion_esperada']}

Resumen profesional:
{fila['resumen_profesional']}

Formación académica:
{fila['formacion_academica']}

Experiencia laboral:
{fila['experiencia_laboral']}

Habilidades técnicas:
{fila['habilidades_tecnicas']}

Habilidades blandas:
{fila['habilidades_blandas']}

Certificaciones:
{fila['certificaciones']}

Idiomas:
{fila['idiomas']}

Proyectos relevantes:
{fila['proyectos_relevantes']}
"""

# Se eliminan espacios innecesarios al inicio y al final del texto.
    return texto.strip()

# Esta función guarda el texto de un CV en formato PDF.
def guardar_pdf(texto, ruta_pdf):
    doc = SimpleDocTemplate(ruta_pdf, pagesize=letter)
    estilos = getSampleStyleSheet()
    elementos = []

    for bloque in texto.split("\n"):
        if bloque.strip() == "":
            elementos.append(Spacer(1, 8))
        else:
            elementos.append(Paragraph(bloque, estilos["Normal"]))
            elementos.append(Spacer(1, 6))

    doc.build(elementos)


# Esta función guarda el texto de un CV en formato PNG.
def guardar_png(texto, ruta_png):
    ancho, alto = 1200, 1600
    imagen = Image.new("RGB", (ancho, alto), "white")
    draw = ImageDraw.Draw(imagen)

    try:
        fuente_titulo = ImageFont.truetype("DejaVuSans-Bold.ttf", 34)
        fuente_texto = ImageFont.truetype("DejaVuSans.ttf", 26)
    except:
        fuente_titulo = ImageFont.load_default()
        fuente_texto = ImageFont.load_default()

    x, y = 60, 60
    lineas = []

    for parrafo in texto.split("\n"):
        if parrafo.strip() == "":
            lineas.append("")
        else:
            lineas.extend(textwrap.wrap(parrafo, width=70))

    for i, linea in enumerate(lineas):
        fuente = fuente_titulo if i == 0 else fuente_texto
        draw.text((x, y), linea, fill="black", font=fuente)
        y += 42

        if y > alto - 80:
            break

    imagen.save(ruta_png)

# Exportar archivos
archivos_generados = []

for _, fila in df_cvs_export.iterrows():
    texto_cv = construir_texto_cv(fila)

    if fila["formato"] == "PDF":
        nombre_archivo = f"{fila['id']}_{fila['perfil'].replace(' ', '_')}.pdf"
        ruta_archivo = os.path.join(carpeta_cvs, nombre_archivo)
        guardar_pdf(texto_cv, ruta_archivo)

    elif fila["formato"] == "PNG":
        nombre_archivo = f"{fila['id']}_{fila['perfil'].replace(' ', '_')}.png"
        ruta_archivo = os.path.join(carpeta_cvs, nombre_archivo)
        guardar_png(texto_cv, ruta_archivo)

    archivos_generados.append({
        "id": fila["id"],
        "perfil": fila["perfil"],
        "nivel": fila["nivel"],
        "formato": fila["formato"],
        "archivo": ruta_archivo
    })

# Se convierte la lista de archivos generados en un DataFrame.
df_archivos = pd.DataFrame(archivos_generados)

# Se muestra la tabla completa de archivos generados.
df_archivos.head(20)
```

## 7.1 Extracción de texto desde documentos PDF y PNG

Para construir la base textual de candidatos, el sistema extrae información de los documentos generados en la base de conocimiento. En el caso de los archivos PDF se recupera el texto directamente mediante `pypdf`; en el caso de los archivos PNG se utiliza OCR mediante `pytesseract`.

Esta etapa permite integrar documentos de formatos distintos dentro de una misma representación textual, lo cual es necesario para construir posteriormente la base vectorial y realizar la búsqueda semántica.

En esta sección se implementa el sistema inteligente para analizar los CVs sintéticos y recomendar los 5 candidatos más adecuados para la vacante seleccionada.

La arquitectura del sistema se compone de las siguientes etapas:

| Etapa | Descripción |
|---|---|
| Entrada documental | Se utilizan los 20 CVs sintéticos generados previamente, en formato PDF y PNG. |
| Extracción de texto | Se extrae texto directamente desde los PDF y mediante OCR desde los PNG. |
| Base textual | Se construye una tabla con el contenido textual de cada CV. |
| Representación vectorial | Se generan embeddings semánticos para cada CV y para la vacante. |
| Búsqueda semántica | Se calcula la similitud entre la vacante y cada CV. |
| Ranking | Se ordenan los candidatos según su similitud semántica con la vacante. |
| Selección final | Se obtiene un Top 5 de candidatos recomendados con base en su alineación con el puesto. |

El sistema no sustituye la decisión del reclutador, sino que funciona como una herramienta de apoyo para priorizar perfiles y facilitar la revisión inicial.

```python
# Se actualizan los paquetes disponibles en el entorno de Google Colab.
# Tesseract se utilizará para extraer texto de los CVs guardados como imágenes PNG.
!apt-get -qq update
!apt-get -qq install -y tesseract-ocr tesseract-ocr-spa
!pip -q install pypdf pytesseract sentence-transformers scikit-learn
```

```python
from pypdf import PdfReader
from PIL import Image
import pytesseract
import pandas as pd
import numpy as np
import os

# Esta función extrae texto de un archivo PDF.
def extraer_texto_pdf(ruta_pdf):
    reader = PdfReader(ruta_pdf)
    texto = ""

    for page in reader.pages:
        texto += page.extract_text() or ""

    return texto.strip()

# Esta función extrae texto de un archivo PNG mediante OCR.
def extraer_texto_png(ruta_png):
    imagen = Image.open(ruta_png)
    texto = pytesseract.image_to_string(imagen, lang="spa")
    return texto.strip()

# Esta función decide qué método de extracción usar según el formato del archivo.
def extraer_texto_documento(ruta, formato):
    if formato == "PDF":
        return extraer_texto_pdf(ruta)
    elif formato == "PNG":
        return extraer_texto_png(ruta)
    else:
        return ""

# Se crea una lista vacía para almacenar los documentos procesados.
documentos_procesados = []

for _, fila in df_archivos.iterrows():
    texto_extraido = extraer_texto_documento(fila["archivo"], fila["formato"])

    documentos_procesados.append({
        "id": fila["id"],
        "perfil": fila["perfil"],
        "nivel": fila["nivel"],
        "formato": fila["formato"],
        "archivo": fila["archivo"],
        "texto_extraido": texto_extraido,
        "longitud_texto": len(texto_extraido)
    })

# Se convierte la lista de documentos procesados en un DataFrame.
df_documentos = pd.DataFrame(documentos_procesados)

# Se valida que se hayan procesado 20 documentos.
assert len(df_documentos) == 20, "Deben procesarse exactamente 20 documentos."


# Se valida que todos los documentos tengan texto extraído.
# Si algún documento tiene longitud 0, significa que falló la extracción de texto.
assert (df_documentos["longitud_texto"] > 0).all(), "Hay documentos sin texto extraído."


# Se muestra una tabla de verificación con la longitud del texto extraído.
# Esto permite revisar que tanto PDFs como PNGs fueron procesados correctamente.
df_documentos[
    [
        "id",
        "perfil",
        "nivel",
        "formato",
        "longitud_texto"
    ]
].head(20)
```

## 7.2 Construcción de la base vectorial

Para esta actividad, la base vectorial se implementó como una matriz de embeddings almacenada en memoria, asociada a un DataFrame con los metadatos de cada candidato. En una implementación productiva, esta misma lógica podría migrarse a una base vectorial especializada como FAISS, ChromaDB, Pinecone o Weaviate.

Una vez extraído el texto de los documentos PDF y PNG, se construye una base vectorial de candidatos. Para ello, se utiliza un modelo de embeddings multilingüe capaz de representar semánticamente tanto la vacante como los CVs.

La consulta principal del sistema será la síntesis de la vacante seleccionada, enriquecida con sus competencias clave. Cada CV será comparado contra esa consulta mediante similitud coseno.

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Se construye el texto de consulta de la vacante.
consulta_vacante = (
    vacante["descripcion_sintetica"]
    + " "
    + " ".join(vacante["competencias_clave"])
)

# Se carga un modelo multilingüe de embeddings.
modelo_embeddings = SentenceTransformer(
    "sentence-transformers/paraphrase-multilingual-MiniLM-L12-v2"
)

textos_cvs = df_documentos["texto_extraido"].tolist()

embeddings_cvs = modelo_embeddings.encode(
    textos_cvs,
    show_progress_bar=True,
    normalize_embeddings=True
)

embedding_vacante = modelo_embeddings.encode(
    [consulta_vacante],
    normalize_embeddings=True
)

# Se calcula la similitud coseno entre la vacante y cada CV.
similitudes = cosine_similarity(embedding_vacante, embeddings_cvs)[0]

df_ranking = df_documentos.copy()

df_ranking["similitud_semantica"] = similitudes
df_ranking["similitud_porcentaje"] = (
    df_ranking["similitud_semantica"] * 100
).round(2)

df_ranking = df_ranking.sort_values(
    "similitud_semantica",
    ascending=False
).reset_index(drop=True)

# Se muestra el Top 10 del ranking semántico inicial.
df_ranking[
    [
        "id",
        "perfil",
        "nivel",
        "formato",
        "similitud_porcentaje"
    ]
].head(10).style.format({
    "similitud_porcentaje": "{:.2f}"
})
```

## 7.4 Ajuste técnico del ranking semántico

El ranking semántico inicial permite identificar candidatos cuyo contenido textual es parecido a la vacante. Sin embargo, para una posición de nivel Lead, la similitud textual no es suficiente. Es necesario considerar también experiencia, nivel profesional, liderazgo técnico, MLOps, cloud computing, CI/CD, sistemas distribuidos, pipelines de datos y uso de frameworks de machine learning.

Por esta razón, se construyó un ranking final ponderado que combina la similitud semántica con criterios técnicos derivados de los requisitos de la vacante.

```python
# Construir tabla auxiliar con información completa de cada CV
df_info_cvs = df_cvs.merge(
    df_diseno_cvs[["id", "experiencia_anios", "alineacion_esperada"]],
    on="id",
    how="left"
)

# Unir ranking semántico con información completa de los CVs
df_ranking_final = df_ranking.merge(
    df_info_cvs[
        [
            "id",
            "experiencia_anios",
            "alineacion_esperada",
            "resumen_profesional",
            "experiencia_laboral",
            "habilidades_tecnicas",
            "habilidades_blandas",
            "certificaciones",
            "proyectos_relevantes"
        ]
    ],
    on="id",
    how="left"
)

# Esta función asigna un puntaje según los años de experiencia del candidato.
def puntaje_experiencia(anios):
    if anios >= 8:
        return 100
    elif anios >= 6:
        return 90
    elif anios >= 4:
        return 75
    elif anios >= 2:
        return 50
    else:
        return 30


# Esta función asigna un puntaje según el nivel profesional del candidato.
def puntaje_nivel(nivel):
    if nivel == "Senior":
        return 100
    elif nivel == "Intermedio":
        return 75
    else:
        return 40

def contiene(texto, palabras):
    texto = texto.lower()
    return any(palabra.lower() in texto for palabra in palabras)

# Esta función calcula un puntaje técnico para cada candidato.
def puntaje_tecnico(fila):
    texto = " ".join([
        str(fila["resumen_profesional"]),
        str(fila["experiencia_laboral"]),
        str(fila["habilidades_tecnicas"]),
        str(fila["habilidades_blandas"]),
        str(fila["certificaciones"]),
        str(fila["proyectos_relevantes"])
    ]).lower()

    puntaje = 0

    # Competencias de machine learning.
    if contiene(texto, ["machine learning", "ml", "modelos", "predictivo", "clasificación"]):
        puntaje += 15

    # Competencias de MLOps y ciclo de vida de modelos.
    if contiene(texto, ["mlops", "mlflow", "kubeflow", "monitoreo", "reentrenamiento", "versionar", "desplegar", "despliegue", "desplegado"]):
        puntaje += 15

    # Experiencia en nube.
    if contiene(texto, ["aws", "azure", "google cloud", "gcp", "cloud"]):
        puntaje += 12

    # Sistemas distribuidos y procesamiento a gran escala.
    if contiene(texto, ["spark", "kafka", "sistemas distribuidos", "distribuido", "databricks"]):
        puntaje += 12

    # Construcción de pipelines de datos.
    if contiene(texto, ["pipeline", "pipelines", "etl", "airflow", "feature store"]):
        puntaje += 12

    # Uso de frameworks de machine learning.
    if contiene(texto, ["pytorch", "tensorflow", "scikit-learn", "xgboost"]):
        puntaje += 10

    # Prácticas de integración y despliegue continuo.
    if contiene(texto, ["ci/cd", "github actions", "pruebas automatizadas", "despliegue continuo"]):
        puntaje += 10

    # Responsible AI, explainability y gobernanza de modelos.
    if contiene(texto, ["responsible ai", "explainable ai", "gobernanza", "explicable"]):
        puntaje += 7

    # Liderazgo técnico y arquitectura.
    if contiene(texto, ["liderazgo", "mentoría", "lead", "arquitectura", "stakeholders"]):
        puntaje += 7

    # Se devuelve el puntaje técnico total del candidato.
    return puntaje

# Se calcula el puntaje por años de experiencia.
df_ranking_final["puntaje_experiencia"] = df_ranking_final["experiencia_anios"].apply(puntaje_experiencia)

# Se calcula el puntaje por nivel profesional.
df_ranking_final["puntaje_nivel"] = df_ranking_final["nivel"].apply(puntaje_nivel)

# Se calcula el puntaje técnico a partir del contenido textual de cada CV.
df_ranking_final["puntaje_tecnico"] = df_ranking_final.apply(puntaje_tecnico, axis=1)

# Se calcula el puntaje final ponderado.
df_ranking_final["puntaje_final"] = (
    df_ranking_final["similitud_porcentaje"] * 0.40 +
    df_ranking_final["puntaje_experiencia"] * 0.20 +
    df_ranking_final["puntaje_nivel"] * 0.15 +
    df_ranking_final["puntaje_tecnico"] * 0.25
).round(2)

# Se ordenan los candidatos de mayor a menor puntaje final.
df_ranking_final = df_ranking_final.sort_values(
    "puntaje_final",
    ascending=False
).reset_index(drop=True)


# Se muestra el Top 10 del ranking final ponderado.
df_ranking_final[
    [
        "id",
        "perfil",
        "nivel",
        "formato",
        "similitud_porcentaje",
        "puntaje_experiencia",
        "puntaje_nivel",
        "puntaje_tecnico",
        "puntaje_final"
    ]
].head(10).style.format({
    "similitud_porcentaje": "{:.2f}",
    "puntaje_final": "{:.2f}"
})
```

## 8.1 Arquitectura implementada

El sistema desarrollado sigue una arquitectura multimodal de recuperación y ranking de candidatos. La entrada está compuesta por 20 CVs sintéticos almacenados en formatos PDF y PNG. | Componente                 | Función                                                                                 |
| -------------------------- | --------------------------------------------------------------------------------------- |
| Vacante seleccionada       | Define los requisitos técnicos y profesionales contra los cuales se comparan los CVs.   |
| Base de CVs sintéticos     | Contiene 20 documentos anónimos con distintos niveles de experiencia y tipos de perfil. |

| Base textual de candidatos | Integra el contenido extraído de todos los CVs en una estructura tabular.               |
| Modelo de embeddings       | Convierte la vacante y los CVs en vectores semánticos comparables.                      |
| Búsqueda semántica         | Calcula la similitud entre la vacante y cada CV mediante similitud coseno.              |

# 9. Ideas adicionales para medir el desempeño del sistema

Además de las métricas utilizadas en la evaluación anterior, el desempeño del sistema podría medirse mediante otros criterios cuantitativos y cualitativos. Estas métricas permitirían evaluar con mayor profundidad si el sistema recupera candidatos adecuados, si el ranking es consistente, si las recomendaciones son explicables y si existen posibles sesgos en el proceso de selección.

| Métrica o criterio                | Descripción                                                                                                                                                                               |
| --------------------------------- | ----------------------------------------------------------------------------------------------------------------------------------------------------------------------------------------- |
| Precision@K                       | Permite medir qué proporción de los primeros K candidatos recomendados son realmente relevantes según un evaluador experto. Por ejemplo, Precision@3 o Precision@10.                      |
| Recall@K                          | Permite medir qué proporción de todos los candidatos relevantes fue recuperada dentro de las primeras K posiciones del ranking.                                                           |
| NDCG                              | Evalúa no solo si los candidatos relevantes aparecen en el ranking, sino también si aparecen en las posiciones correctas. Es útil cuando algunos candidatos son más relevantes que otros. |
| MRR                               | Mide qué tan pronto aparece el primer candidato altamente relevante dentro del ranking. Sería útil si el sistema se usa para reducir rápidamente la carga de revisión.                    |
| Comparación contra juicio experto | Un reclutador técnico o especialista en machine learning podría ordenar manualmente los candidatos y comparar ese orden contra el ranking generado por el sistema.                        |
| Evaluación de explicabilidad      | Se podría revisar si las justificaciones generadas por el sistema son claras, coherentes y basadas en evidencia presente en los CVs.                                                      |
| Auditoría de sesgos               | Se podría analizar si el sistema favorece o penaliza indebidamente ciertos perfiles, niveles de experiencia, tipos de formación o trayectorias profesionales.                             |
| Robustez ante formatos            | Se podría medir si el desempeño cambia entre CVs en PDF y CVs en PNG, especialmente considerando posibles errores de OCR en documentos de imagen.                                         |
| Cobertura de requisitos           | Se podría evaluar cuántos requisitos clave de la vacante aparecen cubiertos por cada candidato recomendado.                                                                               |
| Validación humana posterior       | Se podría analizar si los candidatos recomendados por el sistema también serían seleccionados por un equipo humano en una etapa posterior de revisión.                                    |

Estas mediciones permitirían fortalecer el sistema en una versión futura. En particular, sería importante combinar métricas automáticas con revisión humana, ya que la selección de talento no depende únicamente de similitud textual o coincidencia técnica. También intervienen factores como trayectoria, contexto profesional, capacidad de colaboración, comunicación, liderazgo, adaptación al equipo y potencial de crecimiento.

Una evaluación más completa debería considerar tanto la calidad del ranking como la calidad de las explicaciones. Un buen sistema de apoyo al reclutamiento no solo debe proponer candidatos adecuados, sino también justificar de forma clara por qué los recomienda. De esta manera, el sistema puede ser utilizado como una herramienta transparente, auditable y complementaria al juicio humano.
